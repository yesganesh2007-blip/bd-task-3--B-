**Part 1: Create Bar Plots**

**Purpose:**

To compare sales and profit across categories, regions, products, etc.

**Example question:**

Which category generates maximum profit?

```python
sns.barplot(
    data=df,
    x="Category",
    y="Profit"
)

plt.title("Profit by Category")
plt.show()
```

```python
sns.barplot(
    data=df,
    x="Category",
    y="Sales"
)

plt.title("Sales Distribution by Category")
plt.show()
```

**Part 2: Create Box Plots**

**Purpose:**

Box plots show:

Data distribution
Median value
Outliers
Variation

**Example:**
Profit Distribution

```python
sns.boxplot(
    data=df,
    y="Profit"
)

plt.title("Profit Distribution")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Category",
    y="Profit"
)

plt.title("Profit Variation Across Categories")
plt.show()
```

**Part 3: Analyze Discount vs Profit**

**Purpose:**

Find whether discounts improve sales or reduce profitability.

**Question:**

"At what discount level does profit start decreasing?"

```python
df["Discount"].unique()
```

```python
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Impact of Discount on Profit")
plt.show()
```

**Part 4: Correlation Heatmap**

What is correlation?

Correlation shows the relationship between numerical variables.

```python
numeric_df = df.select_dtypes(
    include="number"
)
```

```python
corr = numeric_df.corr()

corr
```

```python
sns.heatmap(
    corr,
    annot=True
)

plt.title("Correlation Heatmap")
plt.show()
```
